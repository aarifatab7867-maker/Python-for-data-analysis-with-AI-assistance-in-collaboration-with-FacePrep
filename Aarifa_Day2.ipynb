```python
import kagglehub
import os
import pandas as pd

#Download latest version
path=kagglehub.dataset_download("himanshuuike/superstore-sales-dataset")
print("Path to dataset files:",path)

print(os.listdir(path))
df_superstore = pd.read_csv(os.path.join(path,"samplesuperstore.csv"))
print("Dataset loaded successfully")
```

Task-1:Single-Column Groupby Aggregation

Question:Calculate the total sales generated by each Region


```python
#Code
Task1_superstore=df_superstore.groupby("Region")["Sales"].sum().reset_index()
print(Task1_superstore)
```

Task 2:Multi -Column Groupby Aggregation

Question: Calculate total Sales and Profit brokendown by Category and Sub Category

```python
#code
task2_superstore = df_superstore.groupby(['Category','Sub-Category'])[['Sales','Profit']].sum().reset_index()

print(task2_superstore.head(5))
```

"""Task 3-Multi-metric Aggregation with agg()
Question:Compute
-average sales
-Minimum profit
-Maximum profit
for each Product category."""

```python
#code
task3_superstore = df_superstore.groupby('Category').agg(
    Average_Sales=('Sales', 'mean'),
    Minimum_Profit=('Profit', 'min'),
    Maximum_Profit=('Profit', 'max')
).reset_index()

print(task3_superstore)
```

"""Task 4- Sorting Aggregate Summaries
Question: Identify top 5 Sub-Categories generating the highest cumulative Profit."""

```python
#code
task4_superstore = (df_superstore.groupby('Sub-Category')['Profit']
.sum()
.reset_index()
.sort_values(by='Profit', ascending=False)
.head(5))

print(task4_superstore)
```

"""Task 5 - Named Aggreggation


Question:Calculate total Order Count,average Discount rate,and total Profit across Customer Segments."""

```python
#code
task5_superstore = df_superstore.groupby('Segment').agg(
    Total_Order=('Row ID', 'count'),
    Avg_Discount_Rate=('Discount', 'mean'),
    Total_Profit=('Profit', 'sum')
).reset_index()

print(task5_superstore)
```

Task 6:Conditional Group Filtering

Question:Filter the dataset to isolate rows originating only from Cities whose total sales exceed $50,000.

```python
#code
task6_superstore = df_superstore.groupby('City').filter(lambda x: x['Sales'].sum() > 50000)
print("Qualified Cities:",task6_superstore['City'].unique())
print("Retained Rows Ratio:",f"{len(task6_superstore)}/{len(df_superstore)}")

print(task6_superstore)
```

Task 7:Derived Aggregate Calculations

Question:Calculate total sales,total profit,and Profit Margin percentage(Profit/Sales*100)per Ship Mode.

```python
#code
task7_superstore = df_superstore.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()

task7_superstore['Profit_Margin_Percentage'] = (task7_superstore['Total_Profit'] / task7_superstore['Total_Sales']) * 100

print(task7_superstore.sort_values(by='Profit_Margin_Percentage', ascending=False))
```

Task 8- Extremum Selection with nlargest()

Question: Extract the top 5 individual transaction orders based on gross Sales value.

```python
#code
task8_superstore = df_superstore.nlargest(5, 'Sales')[['Order ID','Category','Sales','Profit']]
print(task8_superstore)
```

Task 9- Quantile Aggregation

Question:Find the 75th percentile(Q3) of order Sales across each Product Category.

```python
#code
task9_superstore = df_superstore.groupby('Category')['Sales'].agg(
    Q3_Sales=lambda x: x.quantile(0.75)
).reset_index()
print(task9_superstore)
```

Task 10:Frequency Counting via Groupby

Question:Count total order line items across combinations of Region and Customer Segment.

```python
#code
task10_superstore = (df_superstore.groupby(['Region', 'Segment'])
.size()
.reset_index(name='Order_Count')
.sort_values(by='Order_Count', ascending=False))

print(task10_superstore.head(5))
```

```python
#SUPERSTORE DATASET

import kagglehub

# Download latest version
path = kagglehub.dataset_download("himanshuuike/superstore-sales-dataset")

print("Path to dataset films:",path)
import os

print(os.listdir(path))
import pandas as pd

df_superstore = pd.read_csv(os.path.join(path,"samplesuperstore.csv"))
print("Dataset loaded successfully")
```

TASK 1:Convert Order Date to Datetime and Extract Year,Month,and Day Name

```python
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
df_superstore['Year'] = df_superstore['Order Date'].dt.year
df_superstore['Month'] = df_superstore['Order Date'].dt.month
df_superstore['Day_Name'] = df_superstore['Order Date'].dt.day_name()

display(df_superstore[['Order Date', 'Year', 'Month', 'Day_Name']].head())
```

TASK 2: Clean and Standardize

```python
df_superstore['Customer Name'] = df_superstore['Customer Name'].str.strip().str.upper()

print("Customer Name column after cleaning and converting to uppercase:")
display(df_superstore['Customer Name'].head())
```

TASK 3: Calculate Percentage

```python
df_superstore['Total Category Sales'] = df_superstore.groupby('Category')['Sales'].transform('sum')
df_superstore['Percentage Contribution'] = (df_superstore['Sales'] / df_superstore['Total Category Sales']) * 100

print("Orders with their category sales and percentage contribution:")
display(df_superstore[['Category', 'Sales', 'Total Category Sales', 'Percentage Contribution']].head())
```

TASK 4: Contruct a Pivot table of total sales by region and category.

```python
pivot_table_sales = pd.pivot_table(df_superstore, values='Sales', index='Region', columns='Category', aggfunc='sum')
display(pivot_table_sales)
```

TASK 5: Unpivot Category Sales Data using melt()

```python
unpivoted_sales = pivot_table_sales.reset_index().melt(id_vars=['Region'], var_name='Category', value_name='Sales')
display(unpivoted_sales.head())
```

TASK 6: Merge Regional Target Sales Dataframe with Superstore Data

```python
target_sales_data = {
    'Region': ['Central', 'East', 'South', 'West'],
    'Target Sales': [600000, 750000, 450000, 800000]
}
df_target_sales = pd.DataFrame(target_sales_data)

# Perform an inner merge with df_superstore
df_merged_with_targets = pd.merge(df_superstore, df_target_sales, on='Region', how='inner')

print("Merged DataFrame with Target Sales:")
display(df_merged_with_targets[['Region', 'Sales', 'Target Sales']].head())
```

TASK 7: Categorize Sales into Bins (Low,Medium,High)

```python
bins = [0, df_merged_with_targets['Sales'].quantile(0.33), df_merged_with_targets['Sales'].quantile(0.66), df_merged_with_targets['Sales'].max()]
labels = ['Low', 'Medium', 'High']

df_merged_with_targets['Sales_Category'] = pd.cut(df_merged_with_targets['Sales'], bins=bins, labels=labels, include_lowest=True)

print("Sales categories:")
display(df_merged_with_targets[['Sales', 'Sales_Category']].head())
```

TASK 8: Concatenate Two Quarterly Sales Data Frames

```python
# Sample DataFrames for Q1 and Q2 sales
q1_sales = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar'],
    'Sales': [100, 120, 150],
    'Quarter': ['Q1', 'Q1', 'Q1']
})

q2_sales = pd.DataFrame({
    'Month': ['Apr', 'May', 'Jun'],
    'Sales': [130, 160, 180],
    'Quarter': ['Q2', 'Q2', 'Q2']
})

# Concatenate the two DataFrames vertically
quarterly_sales = pd.concat([q1_sales, q2_sales], ignore_index=True)

print("Concatenated Quarterly Sales Data:")
display(quarterly_sales)
```

TASK 9:Extract Country Code from Order ID using string splitting.

```python
df_superstore['Code_Prefix'] = df_superstore['Order ID'].str.split('-').str[0]

print("DataFrame with 'Code_Prefix' column:")
display(df_superstore[['Order ID', 'Code_Prefix']].head())
```

```python
#SUPERSTORE DATASET

import kagglehub

# Download latest version
path = kagglehub.dataset_download("himanshuuike/superstore-sales-dataset")

print("Path to dataset films:",path)
import os

print(os.listdir(path))
import pandas as pd

df_superstore = pd.read_csv(os.path.join(path,"samplesuperstore.csv"))
print("Dataset loaded successfully")
```

TASK 1:Resample Daily Sales into Weekly Aggregate Sales

```python
import matplotlib.pyplot as plt

# Convert 'Order Date' to datetime and set as index
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
df_superstore_time = df_superstore.set_index('Order Date')

# Resample 'Sales' weekly ('W') using sum aggregation
weekly_sales = df_superstore_time['Sales'].resample('W').sum()

# Plot the line chart
plt.figure(figsize=(12, 6))
weekly_sales.plot(kind='line')
plt.title('Weekly Sales Trend (Superstore)')
plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.grid(True)
plt.tight_layout()
plt.show()
```

TASK 2:Calculate and Plot 7-Day Moving Average of Sales

```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime and set as index for daily operations
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
daily_sales = df_superstore.set_index('Order Date')['Sales'].resample('D').sum()

# Calculate the 7-day rolling mean
daily_sales_rolling_mean = daily_sales.rolling(window=7).mean()

# Plot original daily sales and the 7-day rolling mean
plt.figure(figsize=(14, 7))
plt.plot(daily_sales, label='Daily Sales', alpha=0.7)
plt.plot(daily_sales_rolling_mean, label='7-Day Rolling Mean', color='red')
plt.title('Daily Sales vs. 7-Day Rolling Mean (Superstore)')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

TASK 3: Calculate Period-over-Period Daily Growth using .shift()

```python
# Create a lag-1 column 'Sales_Prev_Day'
daily_sales_df = pd.DataFrame(daily_sales)
daily_sales_df['Sales_Prev_Day'] = daily_sales_df['Sales'].shift(1)

# Calculate daily percentage change in sales
daily_sales_df['Daily_Sales_Pct_Change'] = ((daily_sales_df['Sales'] - daily_sales_df['Sales_Prev_Day']) / daily_sales_df['Sales_Prev_Day']) * 100

print("Daily Sales with previous day's sales and percentage change:")
display(daily_sales_df.head(10))
```

TASK 2: Matplotlib 3D Surface Plot of Synthetic Sales -Profit Surface.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Define the range for x and y
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Create a meshgrid for the x and y coordinates
X, Y = np.meshgrid(x, y)

# Define a function for z (e.g., a simple sinusoidal function for a smooth surface)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a figure and a 3D axes object
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Set labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('Smooth 3D Surface Plot')

# Show the plot
plt.show()
```

TASK 9: Dual-Axis Interactive Graph Objects Chart.

```python
import plotly.graph_objects as go
import pandas as pd
from plotly.subplots import make_subplots

# Ensure 'Order Date' is datetime and aggregate daily sales and profit
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
daily_data = df_superstore.groupby('Order Date').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()

# Create figure with secondary y-axis
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add bar trace for Sales
fig.add_trace(
    go.Bar(
        x=daily_data['Order Date'],
        y=daily_data['Total_Sales'],
        name='Total Sales',
        marker_color='lightblue'
    ),
    secondary_y=False,
)

# Add line trace for Profit
fig.add_trace(
    go.Scatter(
        x=daily_data['Order Date'],
        y=daily_data['Total_Profit'],
        name='Total Profit',
        mode='lines',
        marker_color='red'
    ),
    secondary_y=True,
)

# Set titles and adjust axes
fig.update_layout(
    title_text='Daily Total Sales (Bar) and Total Profit (Line)',
    xaxis_title='Order Date',
    hovermode='x unified'
)

fig.update_yaxes(
    title_text='Total Sales',
    titlefont=dict(color='lightblue'),
    tickfont=dict(color='lightblue'),
    secondary_y=False,
)

fig.update_yaxes(
    title_text='Total Profit',
    titlefont=dict(color='red'),
    tickfont=dict(color='red'),
    secondary_y=True,
)

fig.show()
```
